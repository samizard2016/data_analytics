# Radar Chart / Spider Plot

Radar charts compare multivariate data across categorical axes originating from the center.


```python
import numpy as np
import matplotlib.pyplot as plt

# Define features and data
categories = ['Speed', 'Reliability', 'Comfort', 'Safety', 'Efficiency']
values = [4, 5, 3, 5, 2]

# Split the circle into even angles and repeat the first to close the loop
angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False).tolist()
values += values[:1]
angles += angles[:1]

# Initialize polar plot
fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(projection='polar'))

# Draw the shape
ax.plot(angles, values, color='blue', linewidth=2)
ax.fill(angles, values, color='blue', alpha=0.25)

# Fix the labels
ax.set_thetagrids(np.degrees(angles[:-1]), categories)
plt.show()
```

# Radial Line Plot
Ideal for continuous data or cyclic patterns like hourly, monthly, or environmental metrics over time.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate continuous cyclic data
theta = np.linspace(0, 2 * np.pi, 100)
r = 1 + np.sin(3 * theta) * 0.5

# Initialize polar plot
fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(projection='polar'))
ax.plot(theta, r, color='darkred', lw=2)

# Adjust radial grid limits
ax.set_rmax(2.0)
plt.show()
```

# Radial Bar Chart (Coxcomb)
Useful for showing magnitudes over categorical periods, such as seasonal weather frequency or web traffic

```python
import numpy as np
import matplotlib.pyplot as plt

# Data Setup
categories = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
counts = [12, 19, 3, 5, 2, 3]

# Calculate angles and widths
angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False)
width = (2 * np.pi) / len(categories)

fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(projection='polar'))
bars = ax.bar(angles, counts, width=width, bottom=0.0, edgecolor='white', linewidth=1)

# Style your bars with a colormap
for r, bar in zip(counts, bars):
    bar.set_facecolor(plt.cm.viridis(r / max(counts)))

ax.set_xticklabels(categories)
plt.show()
```

Interactive Alternative: Plotly Express

```python
import plotly.express as px
import pandas as pd

df = pd.DataFrame(dict(
    r=[4, 5, 3, 5, 2],
    theta=['Speed', 'Reliability', 'Comfort', 'Safety', 'Efficiency']
))
fig = px.line_polar(df, r='r', theta='theta', line_close=True)
fig.update_traces(fill='toself')
fig.show()
```
